***<h3><center><u>Objectives</u></center>***</h3>
So the main goal of Word2Vec, simply, is to make accurate vector representations of words.  
This time though, we want these representations to explicitly capture semantic and syntactic meanings of words. So it understands how that **cat** and **dog** are probably similar semantically. But **bigger** and **smaller** syntactically too.

So how? I mean, how can we test that? Perhaps simple similarity measures, even though were and still useable and effective, can't really tell us much about that semantic and syntactic thing. It can definitely indicate the **cat** and **dog**, but for **bigger** and **smaller**, maybe you might misinterpret a high similarity between them?  

But since we will be dealing with vectors, shouldn't, or wouldn't here, I'm not sure, be able to do simple algebraic operations on them?
So for example, assuming vector representations here, we should be able to do this operation:
$$\text{Smaller} - \text{Small} + \text{Big} = \text{Bigger}$$
Such that, we will search with output vector our lookup table of vector representation using *cosine smiliarity*, and find it most similar to "Bigger" vector.

Another example could be the relation between countries and their capitals, so ("Egypt", "Cairo") would be similar to ("England", "London").  
And so on...

***<h3><center><u>Related Works & Approaches</u></center>***</h3>
They discussed several approaches and archictectures in language models that to produced vector representations too, some NN-based architectures include:
1.  NNLM (Neural Network Language Model), the one I discussed and implemented in the other notebook.
2.  RNNLM (Recurrent Neural Network Language Model), haven't discussed that yet honestly but soon.

The main issue with these models are that they aren't really feasible to train on datasets bigger than few hundred millions.  
Why? Mainly due to the computational complexity introduced from the hidden activation layers, and less dominantly in some archictures, the softmax output layer which calculates the probability distribution over all words in the vocabulary $V$.  
But the Softmax layer of dimension $|V|$ has some good optimizations:
1. Using the logits directly from the model (e.g. for *NNLM*), thus not using the softmax layer at all.
2. Keeping the softmax layer (***Hierarchical Softmax*** (Morin, F., Bengio, Y., 2005)), but introducing the words in a binary tree structure:
    * Thus decreasing the the output complexity to $log_2(V)$ instead of $V$.
    * We can also use *Huffman Encoding* to produce the binary tree. In the case of text, we can use the frequency of the words for the encoding. Thus making the output complexity to be $log_2(Unigram(V))$.
